2. Write a Python program to count vowels and consonants in a string.
*   Input Programming
*   Expected Output Vowels = 3, Consonants = 8

```python
c = 'Programming'
vowels = 0
consonants = 0

for char in c:
  if char.lower() in 'aeiou':
    vowels += 1
  else:
    consonants += 1
# this is single word so not checking if char is alphabate
print('Vowels = ', vowels, ',', 'Consonants = ', consonants)

```

```python
c = 'There are 264 cars.'
vowels = 0
alphabates = 0
numbers = 0

for char in c:
  if char.lower() in 'aeiou':
    vowels += 1
  elif char.isalpha():
    alphabates += 1
  elif char.isalnum():
    numbers += 1

print('Vowels = ', vowels)
print('Alphabates = ', alphabates)
print('Numbers = ', numbers)
```

3. Write a Python program to check whether a string is a palindrome or not.
*   Input madam
*   Expected Output The string is a palindrome.

-> same word from left to right or right to left
  ex. madam, radar

```python
d = 'madam'

if d == d[::-1]:
    print('The string is a palindrome.')
else:
    print('The string is not a palindrome.')
```

```python
text = input("Enter a string : ")

if text == text[::-1]:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

10. Write a Python program to find the frequency of each character in a string.
*   Input aabcc
*   Expected Output a: 2, b: 1, c: 2

```python
l = 'aabbbbcccde'

# create dictionary
frequency = {}

for char in l:
    if char in frequency:
        frequency[char] = frequency[char] + 1   # if same char comes it will add +1
    else:
        frequency[char] = 1   # for new char

print(frequency)
```

14. Write a Python program to find the first non-repeating character in a string.

*   Input= aabbcddee

*   Expected Output= First non-repeating character = c



```python
p = 'aabbcddee'

for char in p:
  if p.count(char) == 1:
    print('First non-repeating character = ', char)
```

```python
# to get repeated distinct characters in string
repeating = ''

for char in p:
  if p.count(char) > 1 and char not in repeating:
    repeating = repeating + char

print('Repeated characters = ', repeating)
```

15. Write a Python program to remove duplicate characters from a string while preserving order.
*   Input programming
*   Expected Output Unique characters = progamin




```python
q = 'programming'

uniq = ''

for char in q:
  if char not in uniq:
    uniq = uniq + char

print('Unique characters = ', uniq)
```

16. Write a Python program to count uppercase letters, lowercase letters, digits, and special characters in a
string.
*  Input LihaTech@2026
*  Expected Output Uppercase = 2, Lowercase = 6, Digits = 4, Special = 1

```python
r = 'LihaTech@2026  ! #$%'
uppercase = 0
lowercase = 0
digits = 0
special = 0
space = 0

for char in r:
    if char.isupper():
        uppercase = uppercase + 1
    elif char.islower():
        lowercase = lowercase + 1
    elif char.isdigit():
        digits = digits + 1
    elif char == ' ' :
        space = space + 1
    else:
        special = special + 1

print("Uppercase =", uppercase, ", Lowercase =", lowercase, ", Digits =", digits, ", Special =", special, ", Space =", space)
```
