5. Write a Python program to convert a tuple into a list and a list into a tuple.
*   Input Tuple = (1, 2, 3)
List = [4, 5, 6]
*   Expected Output Tuple to List = [1, 2, 3]
List to Tuple = (4, 5, 6)

```python
T = (1, 2, 3)
L = [4, 5, 6]
# Convert tuple into list
new_list = list(T)

# Convert list into tuple
new_tuple = tuple(L)

print("Tuple to List =", new_list)
print("List to Tuple =", new_tuple)
```

13. Write a Python program to remove duplicate values from a tuple.
* Input (1, 2, 2, 3, 1, 4)
* Expected Output (1, 2, 3, 4)

```python
m = (1, 2, 2, 3, 3, 1, 4, 2)

tuple(set(m))  #using set to remove duplicates and again convert into tuple
#set() does not guarantee the original order
```

```python
m = (1, 2, 2, 3, 3, 1, 4, 2)

tuple(dict.fromkeys(m))   # dict dictionary to remove duplicates and also keep the same order of elements
```

```python
# using for loop
m = (1, 2, 2, 3, 3, 1, 4, 2)
n =()
for item in m:
    if item not in n:
        n = n + (item,)
n
```

14. Write a Python program to create a nested tuple and access its inner elements.
* Input (1, 2, (10, 20, 30))
* Expected Output Inner tuple = (10, 20, 30)
Second inner element = 20

```python
o = (1,2)
p =(10,20,30)

#o + p   # output is (1, 2, 10, 20, 30)

o = o + (p,)    # output is (1, 2, (10, 20, 30))
print("Nested tuple =", o)

print("Inner tuple =", o[2])
print("Second inner element =", o[2][1])
```

15. Write a Python program to swap two tuples.
* Input Tuple A = (1, 2)
Tuple B = (3, 4)
* Expected Output Tuple A = (3, 4)
Tuple B = (1, 2)

```python
A = (1, 2)
B = (3, 4)

A, B = B, A   # direct assignment of objects
print("Tuple A =", A)
print("Tuple B =", B)
```
