11. Write a Python program to sort a list of tuples by the second value using a lambda function.
Input items = [(1, 30), (2, 10), (3, 20)]
Expected Output [(2, 10), (3, 20), (1, 30)]


```python
items = [(1, 30), (2, 10), (3, 20)]
sorted_items = sorted(items, key=lambda x: x[1])  #x[0] = 1st value and x[1] = second value
print(sorted_items)
```

16. Write a Python program to convert Celsius values to Fahrenheit using map() and lambda.

1. Write a Python program to sort a list of employee dictionaries by age using a lambda key.
Input employees = [{'name':'Asha','age':29}, {'name':'Ravi','age':24}, {'name':'Neha','age':31}]
Expected Output ['Ravi', 'Asha', 'Neha']


```python
employees = [{'name':'Asha','age':29}, {'name':'Ravi','age':24}, {'name':'Neha','age':31}]
sorted_employees = sorted(employees, key=lambda x: x['age'])
sorted_employees
```

```python
# find out max age record
max_age = max(employees, key=lambda x: x['age'])
max_age
```

2. Write a Python program to sort a list of names by the last character using a lambda function.
Input names = ['Asha', 'Ravi', 'Neha', 'Aman']
Expected Output ['Asha', 'Neha', 'Ravi', 'Aman']


```python
names = ['Asha', 'Ravi', 'Neha', 'Aman']
sorted_names = sorted(names, key=lambda x: x[-1])  # same as reverse indexing -1
sorted_names
```

3. Write a Python program to apply a 10% discount to every price using map() and lambda.
Input prices = [500, 1000, 1500, 2000]
Expected Output [450.0, 900.0, 1350.0, 1800.0]


```python
prices = [500, 1000, 1500, 2000]
disc_prices = list(map(lambda x: x - (x * 0.10), prices))
disc_prices
```

6. Write a Python program to filter palindrome words using lambda.
Input words = ['level', 'python', 'madam', 'cloud', 'radar']
Expected Output ['level', 'madam', 'radar']


```python
words = ['level', 'python', 'madam', 'cloud', 'radar']
palindrom = list(filter(lambda x: x == x[::-1], words))
palindrom
```

10. Write a Python program to sort student records by marks in descending order and then by name in
ascending order using lambda.
Input students = [('Ravi', 85), ('Asha', 92), ('Neha', 85), ('Aman', 92)]
Expected Output  [('Aman', 92), ('Asha', 92), ('Neha', 85), ('Ravi', 85)]

```python
students = [('Ravi', 85), ('Asha', 92), ('Neha', 85), ('Aman', 92)]
sorted_students = sorted(students, key=lambda x: (-x[1], x[0]))    # multiple conditions
sorted_students

# x[0] = student name , ascending
# x[1] = marks
# -x[1] = makes marks (-) used for descending
```
