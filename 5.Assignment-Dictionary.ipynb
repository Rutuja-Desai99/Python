9. Write a Python program to count the frequency of characters in a string using a dictionary.
* Input aabcc
* Expected Output {'a': 2, 'b': 1, 'c': 2}

```python
h = "aabcc"
count = {}

for item in h:
  if item in count:
    count[item] += 1
  else:
    count[item] = 1

print(count)

#for char, count in count.items():
#    print(char, ":", count)
```

Q. Problem Statement: Sort employee records based on salary using lambda. Solution: employees = [ {"name": "Amit", "salary": 45000}, {"name": "Neha", "salary": 60000}, {"name": "Ravi", "salary": 38000} ]

```python
emp = [ {"name": "Amit", "salary": 45000}, {"name": "Neha", "salary": 60000}, {"name": "Ravi", "salary": 38000} ]
sorted(emp, key=lambda x: x["salary"])   # ascending order
```

```python
sorted(emp, key=lambda x: x["salary"], reverse=True)   #descending
```

13. Write a Python program to find the key with the maximum value in a dictionary.
* Input {"A": 75, "B": 92, "C": 88}
*  Expected Output Key with maximum value = B

```python
k = {"A": 75, "B": 92, "C": 88}
max(k, key=k.get)    #max() function with key=dict.get
```

16. Write a Python program to count the frequency of words in a sentence using a dictionary.
* Input python is easy and python is powerful
* Expected Output {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}

```python
st = "python is easy and python is powerful"
cnt = {}
for word in st.split():
  if word in cnt:
    cnt[word] += 1
  else:
    cnt[word] = 1

print(cnt)
```
