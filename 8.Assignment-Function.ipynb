5. Write a Python function to count the frequency of each element in a list and return a dictionary.
Input items = ['A', 'B', 'A', 'C', 'B', 'A']
Expected Output {'A': 3, 'B': 2, 'C': 1}

```python
def Frequency(items):
  freq = {}
  for i in items:
    if i in freq:
      freq[i] += 1
    else:
      freq[i] = 1
  return freq

Frequency(['A', 'B', 'A', 'C', 'B', 'A'])
```

6. Write a recursive Python function to find the sum of digits of a number.
Input number = 5832
Expected Output Sum of digits = 18

```python
def SumOfDigits(number):
    total = 0
    while number > 0:
        digit = number % 10    # finds last digit
        total += digit         # add in total
        number = number // 10  # removes digit
    return total

#SumOfDigits(5832)
print("Sum of digits =", SumOfDigits(5832))
```

7. Write a recursive Python function to return the nth Fibonacci number.
Input n = 8
Sequence starts with 0, 1
Expected Output Fibonacci number = 21

```python
def Fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return Fibonacci(n - 1) + Fibonacci(n - 2)

#Fibonacci(3)   # start with 0 so o/p is 2
print("Fibonacci number =", Fibonacci(8))
```

8. Write a Python function to calculate income tax using slabs: 0% up to 300000, 5% for the next 300000,
and 10% for the remaining amount.
Input income = 800000
Expected Output Tax = 35000.0


```python
def CalculateTax(income):
    if income <= 300000:
        tax = 0
    elif income <= 600000:
        tax = (income - 300000) * 0.05
    else:
        tax = (300000 * 0.05) + ((income - 600000) * 0.10)
    return tax

#income = float(input("Enter your income: "))
print("Income Tax:", CalculateTax(800000))
```

10. Write a Python function to validate a password and return both status and reason. Require at least 8
characters, one uppercase letter, and one digit.
Input password = Python123
Expected Output Valid = True

Reason = Password accepted

```python
def ValidatePassword(password):
    if len(password) < 8:
        return "Invalid: Password must be at least 8 characters"
    if not any(ch.isupper() for ch in password):
        return "Invalid: Must contain 1 uppercase letter"
    if not any(ch.islower() for ch in password):
        return "Invalid: Must contain 1 lowercase letter"
    if not any(ch.isdigit() for ch in password):
        return "Invalid: Must contain 1 digit"
    else:
        return "Password is Valid"

    #if not any(not ch.isalnum() for ch in password):
    #    return "Invalid: Must contain 1 special character"

#password = input("Enter password: ")
password = "Python123"
print(ValidatePassword(password))
```
