2. Write a Python program to find the sum and average of list elements.
*   Input [10, 20, 30, 40]
*   Expected Output Sum = 100, Average = 25.0

```python
b = [10, 20, 30, 40]
sum = 0
for item in b:
  sum += item
print("Sum = ", sum, ", Average = ", sum/len(b))
```

4. Write a Python program to count even and odd numbers in a list.
*   Input [1, 2, 3, 4, 5, 6]
*   Expected Output Even count = 3, Odd count = 3

```python
c = [1, 2, 3, 4, 5, 6]
Even = 0
Odd = 0

for item in c:
  if item%2 == 0:
    Even = Even + 1
  else:
    Odd = Odd + 1
print("Even count = ", Even, ", Odd count = ", Odd)
```

5. Write a Python program to remove duplicate elements from a list.
*   Input [1, 2, 2, 3, 1, 4]

6. Write a Python program to reverse a list without using the reverse() method.
*   Input [10, 20, 30, 40]
*   Expected Output [40, 30, 20, 10]

```python
f = [10, 20, 30, 40]
# using reverse method -  f.reverse()
#len(f) = 4  #index strats with o,1,2,3

reverse = []
for i in range(len(f) - 1, -1, -1):
    reverse.append(f[i])

print("Original list:", f)
print("Reversed list:", reverse)
```

```python
numbers = [1, 2, 3, 4, 5]

left = 0
right = len(numbers) - 1

while left < right:
    numbers[left], numbers[right] = numbers[right], numbers[left]

    left = left + 1
    right = right - 1

print(numbers)
```

13. Write a Python program to separate positive and negative numbers from a list.
* Input [-5, 3, -2, 7, 0, -1]
* Expected Output Positive = [3, 7]
Negative = [-5, -2, -1]

```python
n = [-5, 3, -2, 7, 0, -1]
positive = []
negative = []

for item in n:
  if item > 0:
    positive.append(item)
  elif item < 0:
    negative.append(item)
print("Positive = ", positive)
print("Negative = ", negative)
```

14. Write a Python program to find common elements between two lists.
* Input List 1 = [1, 2, 3, 4]
List 2 = [3, 4, 5, 6]
* Expected Output Common elements = [3, 4]

```python
o = [1, 2, 3, 4]
p = [3, 4, 5, 6]
com = []

for item in o:
  if item in p:
    com.append(item)
print("Common elements = ", com)
```

```python
# other way using set
cmn = list(set(o) & set(p))
cmn
```

16. Write a Python program to rotate a list by n positions.
*  Input List = [1, 2, 3, 4, 5]
n = 2
*  Expected Output Rotated list = [3, 4, 5, 1, 2]

```python
r = [1, 2, 3, 4, 5]
n = 2

for item in range(n):
    temp = r.pop(0)
    r.append(temp)
print("Rotated list = ", r)
```
