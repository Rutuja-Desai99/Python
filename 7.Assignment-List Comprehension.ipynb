appearance order.

2. Write a Python program to generate all pairs whose sum is equal to a given target.
Input numbers = [1, 2, 3, 4, 5, 6]

target = 7
Expected Output [(1, 6), (2, 5), (3, 4)]


```python
numbers = [1, 2, 3, 4, 5, 6]
target = 7

pairs = [(numbers[i], numbers[j])
         for i in range(len(numbers))
         for j in range(i + 1, len(numbers))
         if numbers[i] + numbers[j] == target]

print(pairs)
```

9. Write a Python program to generate all unique Pythagorean triples where 1 <= a < b < c <= 20.
Input Range = 1 to 20
Expected Output [(3, 4, 5), (5, 12, 13), (6, 8, 10), (8, 15, 17), (9, 12, 15), (12, 16, 20)]


```python
n = 21

triples = [(a, b, c)
           for a in range(1, n)
           for b in range(a, n)
           for c in range(b, n)
           if a*a + b*b == c*c]
triples
```
